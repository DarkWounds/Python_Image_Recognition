```
from ultralytics import YOLO
model = YOLO('yolov8n.pt')
model.train(data='/content/dataset/data.yaml', epochs=50, imgsz=640)
```

```
# Test pe pozele din setul de test
best = YOLO('/content/runs/detect/train/weights/best.pt')
best.val(data='/content/dataset/data.yaml', split='test')
res = best.predict('/content/dataset/test/images', save=True, conf=0.4)
print('Pozele cu predicții sunt în:', res[0].save_dir)
```
